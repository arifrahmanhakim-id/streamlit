Memuat dataset

https://www.kaggle.com/code/dhainjeamita/breast-cancer-dataset-classification/input

```python
import pandas as pd
df = pd.read_csv('breast_cancer.csv')
df.head()
```

Memeriksa Data yang kosong

```python
df.isna().sum()
```

Data Reduction

```python
# menghapus kolom yang tidak dibutuhkan
df.drop(df.columns[[-1, 0]], axis=1,
        inplace=True)
df.info()
```

Membagi data X dan y

```python
#X adalah variabel input, y adalah variabel target
semua = list(df.columns[1:31]) 
X = df.loc[:,semua] 
y = df[['diagnosis']] 
print(X.shape) 
print(y.shape)
```

Memilah data training dan testing

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,
test_size=0.3, random_state=0)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

Membuat Model

```python
from sklearn import svm
#create a classifier
model = svm.SVC(kernel="linear")
#train the model
modelfit = model.fit(X_train,y_train)
accuracy = modelfit.score(X_test, y_test)
#predict the response
pred = model.predict(X_test)
print(accuracy)
```

Mengevaluasi Model

```python
from sklearn.metrics import classification_report 
print(classification_report(y_test, pred))
```

Menyimpan Model

```python
# Import model ke File pickle
import pickle
with open('svm_pickle','wb') as r:
    pickle.dump(modelfit,r)
```

Memuat dataset

https://www.kaggle.com/datasets/uciml/iris

```python
import pandas as pd
#TAHAPAN 1 IMPORT DATA
df = pd.read_csv('Iris.csv') 
df.head()
```

Melihat Nlai yang kosong

```python
# Melihat missing value
df.isnull().sum()
```

Data Reduction

```python
# Memisahkan variabel dan class
X = df.drop(['Id', 'Species'], axis=1) 
y = df[['Species']]
X.head()
```

Memilah Data Training dan Testing

```python
# Split Data Training = 70% dan Data Testing 30%
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X, 
y, test_size = 0.3, random_state=0)
print(X_train.shape, y_train.shape) 
print(X_test.shape, y_test.shape)
```

Membuat Model Random Forest

```python
# Training Model Random Forest
from sklearn.ensemble import RandomForestClassifier 
#create a classifier
model = RandomForestClassifier(n_estimators=40) 
#train the model
modelfit = model.fit(X_train, y_train.values.ravel()) 
accuracy = modelfit.score(X_test, y_test.values.ravel())
#predict the response
y_pred = model.predict(X_test)
print(accuracy)
```

Mengevaluasi Model

```python
# Evaluasi model dengan confusion matrix
from sklearn.metrics import classification_report 
print(classification_report(y_test, y_pred))
```

Import model ke File pickle

```python
import pickle
with open('randomforet_pickle','wb') as r: 
    pickle.dump(modelfit,r)
```
